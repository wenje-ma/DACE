# 2 用于描述计算机模拟器输出的随机过程模型

> **图 2.1**：在区间 $[-1,+1]$ 上对随机函数 $Y(x)=b_0+b_1x+b_2x^2$ 进行 $10$ 次抽样，其中 $b_0$、$b_1$ 和 $b_2$ 相互独立，且均服从标准正态分布 $\mathrm{N}(0,1.0)$

```r
if(!dir.exists("data"))dir.create("data")
if(!file.exists("data/2.1-plot.RData")){
  set.seed(1)
  n=10
  x=seq(-1,1,length.out=301) # 301x1
  b=matrix(rnorm(3*n),nrow=n,ncol=3) # 10x3
	save(n,x,b,file="data/2.1-plot.RData")
}

load("data/2.1-plot.RData")

options(repr.plot.width=4,repr.plot.height=4)
plot(x,rep(0,length(x)),type="n",xlim=c(-1,1),ylim=c(-4,4),xlab="x",ylab="y(·)")
for(i in 1:n){
	y=b[i,1]+b[i,2]*x+b[i,3]*x^2 # 301x1
  lines(x,y)
}
abline(v=0)
```

> **图 2.2**：左侧面板：对于平稳过程，每个箭头的尖端与尾端具有相同的协方差。右侧面板：对于各向同性过程，原点和圆上所有点具有相同的相关性

```r
if(!dir.exists("data"))dir.create("data")
if(!file.exists("data/2.2-plot.RData")){
	x=seq(-5,5,length.out=301) # 301x1
  theta=seq(0,2*pi,length.out=301) # 301x1
  save(x,theta,file="data/2.2-plot.RData")
}

load("data/2.2-plot.RData")

options(repr.plot.width=8,repr.plot.height=4)
par(mfrow=c(1,2))
plot(NULL,xlim=c(-5,5),ylim=c(-4,4),xlab="",ylab="")
abline(h=0,v=0)
lines(x,-x)
lines(x,-x+2)
arrows(x0=c(0,1,-1),y0=c(0,-1,1),x1=c(1,2,0),y1=c(1,0,2),length=0.1)
plot(NULL,xlim=c(-5,5),ylim=c(-4,4),xlab="",ylab="")
abline(h=0,v=0)
xc=4*cos(theta)
yc=4*sin(theta)
lines(xc,yc)
ang=c(pi/4,3*pi/4,5*pi/4,7*pi/4)
xp=4*cos(ang)
yp=4*sin(ang)
arrows(0,0,xp,yp,length=0.1)
par(mfrow=c(1,1))
```

```r
if(!dir.exists("data"))dir.create("data")
if(!file.exists("data/2.3-plot.RData")){
  set.seed(1)
  gp=function(x,l){
    n=length(x) # x
    Sigma=exp(-outer(x,x,"-")^2/(2*l^2)) # (x)x(x)
    t(chol(Sigma+diag(n)*1e-9))%*%rnorm(n) # (x)x(1)
  }
  x=seq(0,1,length.out=301) # 301x1
  n=5
  save(gp,x,n,file="data/2.3-plot.RData")
}

load("data/2.3-plot.RData")

options(repr.plot.width=8,repr.plot.height=8)
par(mfrow=c(2,1))
plot(NULL,xlim=c(0,1),ylim=c(-3,3),xlab="",ylab="")
for(i in 1:n){
  y=gp(x,l=0.8)
  lines(x,y,lty=2)
}
plot(NULL,xlim=c(0,1),ylim=c(-3,3),xlab="",ylab="")
for(i in 1:n){
  y=gp(x,l=0.4)
  lines(x,y,lty=2)
}
par(mfrow=c(1,1))
```

> **图 2.4**：一个函数 $y(x_1,x_2)$，从直观上看与平稳模型不一致

```r
if(!dir.exists("data"))dir.create("data")
if(!file.exists("data/2.4-plot.RData")){
  set.seed(1)
  gp2d=function(x1,x2,l){
    grid=expand.grid(x1=x1,x2=x2) # {(x1)x(x2)}x{2}
    D1=outer(grid$x1,grid$x1,"-") # {(x1)x(x2)}x{(x1)x(x2)}
    D2=outer(grid$x2,grid$x2,"-") # {(x1)x(x2)}x{(x1)x(x2)}
    Dist=sqrt(D1^2+D2^2) # {(x1)x(x2)}x{(x1)x(x2)}
    Sigma=exp(-Dist^2/(2*l^2))+diag(nrow(Dist))*1e-9 # {(x1)x(x2)}x{(x1)x(x2)}
    z=as.vector(t(chol(Sigma))%*%rnorm(nrow(Dist))) # {(x1)x(x2)}x{1}
    matrix(z,nrow=length(x1),ncol=length(x2)) # (x1)x(x2)
  }
  x=y=seq(0,1,length.out=50)
  z=gp2d(x,y,l=0.1)
  save(x,y,z,file="data/2.4-plot.RData")
}

load("data/2.4-plot.RData")

options(repr.plot.width=4,repr.plot.height=4)
persp(x,y,z,theta=30,phi=20,xlab="",ylab="",zlab="")
```

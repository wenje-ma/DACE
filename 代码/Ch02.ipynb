# 2 用于描述计算机模拟器输出的随机过程模型

> **图 2.1**：在区间 $[-1,+1]$ 上对随机函数 $Y(x)=b_0+b_1x+b_2x^2$ 进行 $10$ 次抽样，其中 $b_0$、$b_1$ 和 $b_2$ 相互独立，且均服从标准正态分布 $\mathrm{N}(0,1.0)$

```r
if(!dir.exists("data"))dir.create("data")
if(!file.exists("data/2.1-plot.RData")){
  set.seed(1)
  n=10
  x=seq(-1,1,length.out=301)
  b=matrix(rnorm(3*n),nrow=n,ncol=3)
	save(n,x,b,file="data/2.1-plot.RData")
}
load("data/2.1-plot.RData")
library(svglite)
svglite("../markdown/figures/2.1.svg",width=4,height=4)
plot(x,rep(0,length(x)),type="n",xlim=c(-1,1),ylim=c(-4,4),xlab="x",ylab="y(·)")
for(i in 1:n){
	y=b[i,1]+b[i,2]*x+b[i,3]*x^2
  lines(x,y)
}
abline(v=0)
dev.off()
```

> **图 2.3**：来自两个平稳过程各自的五组抽样结果

```r
if(!dir.exists("data"))dir.create("data")
if(!file.exists("data/2.3-plot.RData")){
  set.seed(1)
  gp=function(x,l){
    n=length(x) # x
    Sigma=exp(-outer(x,x,"-")^2/(2*l^2))
    t(chol(Sigma+diag(n)*1e-9))%*%rnorm(n)
  }
  x=seq(0,1,length.out=301)
  n=5
  save(gp,x,n,file="data/2.3-plot.RData")
}
load("data/2.3-plot.RData")
library(svglite)
svglite("../markdown/figures/2.3.svg",width=8,height=8)
par(mfrow=c(2,1))
plot(NULL,xlim=c(0,1),ylim=c(-3,3),xlab="",ylab="")
for(i in 1:n){
  y=gp(x,l=0.8)
  lines(x,y,lty=2)
}
plot(NULL,xlim=c(0,1),ylim=c(-3,3),xlab="",ylab="")
for(i in 1:n){
  y=gp(x,l=0.4)
  lines(x,y,lty=2)
}
par(mfrow=c(1,1))
dev.off()
```

> **图 2.4**：一个函数 $y(x_1,x_2)$，从直观上看与平稳模型不一致

```r
if(!dir.exists("data"))dir.create("data")
if(!file.exists("data/2.4-plot.RData")){
  set.seed(1)
  gp2d=function(x1,x2,l){
    grid=expand.grid(x1=x1,x2=x2)
    D1=outer(grid$x1,grid$x1,"-")
    D2=outer(grid$x2,grid$x2,"-")
    Dist=sqrt(D1^2+D2^2)
    Sigma=exp(-Dist^2/(2*l^2))+diag(nrow(Dist))*1e-9
    z=as.vector(t(chol(Sigma))%*%rnorm(nrow(Dist)))
    matrix(z,nrow=length(x1),ncol=length(x2))
  }
  x=y=seq(0,1,length.out=50)
  z=gp2d(x,y,l=0.1)
  save(x,y,z,file="data/2.4-plot.RData")
}
load("data/2.4-plot.RData")
library(svglite)
svglite("../markdown/figures/2.4.svg",width=4,height=4)
persp(x,y,z,theta=30,phi=20,xlab="",ylab="",zlab="")
dev.off()
```

> **图 2.5**：相关函数 $R(h|\xi)=\sin(h/\xi)/(h/\xi)$，在区间 $[−1,+1]$ 内，$\xi=1/4\pi$

```r
if(!dir.exists("data"))dir.create("data")
if(!file.exists("data/2.5-plot.RData")){
  R=function(h,xi=1/(4*pi))ifelse(h==0,1,sin(h/xi)/(h/xi))
  h=seq(-1,1,length.out=301)
  y=R(h)
  save(h,y,file="data/2.5-plot.RData")
}
load("data/2.5-plot.RData")
library(svglite)
svglite("../markdown/figures/2.5.svg",width=4,height=4)
plot(h,y,type="l",xlab="h",ylab="sin(4*pi*h)/(4*pi*h)",ylim=c(-0.25,1.05))
abline(h=0)
dev.off()
```

> **图 2.6**：左列：幂指数相关函数 (2.2.10) 中幂次 $p$ 取不同值，对零均值、单位方差且固定 $\xi\equiv1.0$ 的高斯过程的四组采样结果产生的影响；右列：$C(h)=\exp(-|h|^p)$，取值范围 $-2.0\le h\le+2.0$。其中第一行 $p=2.0$，第二行 $p=0.75$，第三行 $p=0.20$

```r
if(!dir.exists("data"))dir.create("data")
if(!file.exists("data/2.6-plot.RData")){
	set.seed(1)
  powexp.corr=function(h,xi,p)exp(-xi*abs(h)^p)
  gp.sample=function(x,xi,p,n.sample=4){
    n=length(x)
    Sigma=outer(x,x,function(a,b)powexp.corr(a-b,xi,p))
    Z=t(chol(Sigma+diag(n)*1e-9))%*%matrix(rnorm(n*n.sample),nrow=n,ncol=n.sample)
    list(x=x,Z=t(Z))
  }
  xi=1.0
  p.vec=c(2.0,0.75,0.20)
  xgrid=seq(0,1,length.out=301)
  hgrid=seq(-2,2,length.out=301)
  gp_list=list()
  rh_list=list()
  for(k in seq_along(p.vec)){
    p=p.vec[k]
    gp_list[[k]]=gp.sample(xgrid,xi=xi,p=p,n.sample=4)
    rh_list[[k]]=powexp.corr(hgrid,xi=xi,p=p)
  }
  save(xi,p.vec,xgrid,hgrid,gp_list,rh_list,file="data/2.6-plot.RData")
}
load("data/2.6-plot.RData")
library(svglite)
svglite("../markdown/figures/2.6.svg",width=5,height=6)
layout.mat=matrix(1:6,nrow=3,ncol=2,byrow=TRUE)
layout(layout.mat,widths=c(2,1),heights=c(1,1,1))
for(k in seq_along(p.vec)){
  gp=gp_list[[k]]
  rh=rh_list[[k]]
  matplot(gp$x,t(gp$Z),type="l",lty=2,col=1,xlab="x",ylab="",ylim=c(-3,3))
  plot(hgrid,rh,type="l",lty=2,xlab="h",ylab="R(h)",ylim=c(0,1))
}
dev.off()
```

> **图 2.7**：左列：幂指数相关函数 (2.2.10) 中变化率参数 $\xi$ 对零均值、单位方差高斯过程的四组采样结果产生的影响，其中固定 $p\equiv2.0$；右列：$C(h)=\exp(-\xi h^2)$，取值范围 $-1.5\le h\le+1.5$。其中 $\xi=4$（第一行），$\xi=16$（第二行），$\xi=100$（第三行）。

```r
if(!dir.exists("data"))dir.create("data")
if(!file.exists("data/2.7-plot.RData")){
	set.seed(1)
  powexp.corr=function(h,xi,p){
    exp(-xi*abs(h)^p)
  }
  gp.sample=function(x,xi,p,n.sample=4){
    n=length(x)
    Sigma=outer(x,x,function(a,b)powexp.corr(a-b,xi,p))
    Z=t(chol(Sigma+diag(n)*1e-9))%*%matrix(rnorm(n*n.sample),nrow=n,ncol=n.sample)
    list(x=x,Z=t(Z))
  }
  p=2.0
  xi.vec=c(4,16,100)
  xgrid=seq(0,1,length.out=301)
  hgrid=seq(-1.5,1.5,length.out=301)
  gp_list=list()
  rh_list=list()
  for(k in seq_along(xi.vec)){
    xi=xi.vec[k]
    gp_list[[k]]=gp.sample(xgrid,xi=xi,p=p,n.sample=4)
    rh_list[[k]]=powexp.corr(hgrid,xi=xi,p=p)
  }
  save(p,xi.vec,xgrid,hgrid,gp_list,rh_list,file="data/2.7-plot.RData")
}
load("data/2.7-plot.RData")
library(svglite)
svglite("../markdown/figures/2.7.svg",width=5,height=6)
layout.mat=matrix(1:6,nrow=3,ncol=2,byrow=TRUE)
layout(layout.mat,widths=c(2,1),heights=c(1,1,1))
for(k in seq_along(xi.vec)){
  gp=gp_list[[k]]
  rh=rh_list[[k]]
  matplot(gp$x,t(gp$Z),type="l",lty=2,col=1,xlab="x",ylab="",ylim=c(-3,3))
  plot(hgrid,rh,type="l",lty=2,xlab="h",ylab="R(h)",ylim=c(0,1))
}
dev.off()
```

> **图 2.8**：左列：三次相关函数 (2.2.12) 中尺度参数 $\psi$ 取不同值时，对零均值、单位方差高斯过程的四组采样结果产生的影响；右列：三次相关函数 $R(h\mid\psi)$，取值范围 $-2.0\le h\le+2.0$。其中 $\psi=10.0$（第一行），$\psi=1.0$（中间行），$\psi=0.5$（最下方一行）

```r
if(!dir.exists("data"))dir.create("data")
if(!file.exists("data/2.8-plot.RData")){
	set.seed(1)
  cubic.corr=function(h,psi){
    ah=abs(h)
    ifelse(ah<=psi/2,1-6*(ah/psi)^2+6*(ah/psi)^3,
  	ifelse(ah<=psi,2*(1-ah/psi)^3,0))
  }
  gp.sample=function(x,psi,n.sample=4){
    n=length(x)
    Sigma=outer(x,x,function(a,b)cubic.corr(a-b,psi)) # (x)x(x)
    Z=t(chol(Sigma+diag(n)*1e-9))%*%matrix(rnorm(n*n.sample),nrow=n,ncol=n.sample) # (x)x1
    list(x=x,Z=t(Z))
  }
  psi.vec=c(10,1,0.5)
  xgrid=seq(0,1,length.out=301)
  hgrid=seq(-2,2,length.out=301)
  gp_list=list()
  rh_list=list()
  for(k in seq_along(psi.vec)){
    psi=psi.vec[k]
    gp_list[[k]]=gp.sample(xgrid,psi,n.sample=4)
    rh_list[[k]]=cubic.corr(hgrid,psi)
  }
  save(psi.vec,xgrid,hgrid,gp_list,rh_list,file="data/2.8-plot.RData")
}
load("data/2.8-plot.RData")
library(svglite)
svglite("../markdown/figures/2.8.svg",width=5,height=6)
layout.mat=matrix(1:6,nrow=3,ncol=2,byrow=TRUE)
layout(layout.mat,widths=c(2,1),heights=c(1,1,1))
for(k in seq_along(psi.vec)){
  gp=gp_list[[k]]
  rh=rh_list[[k]]
  matplot(gp$x,t(gp$Z),type="l",lty=2,col=1,xlab="x",ylab="",ylim=c(-3,3))
  plot(hgrid,rh,type="l",lty=2,xlab="h",ylab="R(h)",ylim=c(0,1))
}
dev.off()
```

> **表 2.1**：从信息丰富的 $[\beta_0,\lambda_Z,\rho]$ 先验分布中抽取 $5$ 个样本，各分量相互独立

```r
if(!dir.exists("data"))dir.create("data")
if(!file.exists("data/2.1-mat.RData")){
  set.seed(1)
  n=5
  beta0=rnorm(n,mean=20,sd=4)
  lambdaZ=rgamma(n,shape=6.25,rate=25)
  sigmaZ=1/sqrt(lambdaZ)
  rho=rbeta(n,shape1=2,shape2=3)
  mat=cbind(beta0,lambdaZ,sigmaZ,rho)
  colnames(mat)=c("beta0","lambdaZ","sigmaZ","rho")
  rownames(mat)=1:n
  save(mat,file="data/2.1-mat.RData")
}
load("data/2.1-mat.RData")
print(round(mat,2))
```

> **图 2.10**：从式 (2.3.2) 中抽取 $5$ 个样本，分别对应表 2.1 中的每一行

```r
if(!dir.exists("data"))dir.create("data")
if(!file.exists("data/2.10-plot.RData")){
	set.seed(1)
  gauss_corr=function(h,rho)rho^(abs(h)^2)
  gp.sample=function(x,rho,sigmaZ,n.sample=1){
    n=length(x)
    Rmat=outer(x,x,function(a,b)gauss_corr(a-b,rho))
    Z=t(chol(sigmaZ^2*Rmat+diag(n)*1e-9))%*%matrix(rnorm(n*n.sample),nrow=n,ncol=n.sample)
    list(x=x,Z=t(Z))
  }
  load("data/2.1-mat.RData")
  xgrid=seq(0,1,length.out=301)
  n_draw=nrow(mat)
  Y_curves=matrix(NA,nrow=n_draw,ncol=length(xgrid))
  for(i in 1:n_draw){
    gp_res=gp.sample(xgrid,rho=mat[i,"rho"],sigmaZ=mat[i,"sigmaZ"],n.sample=1)
    Y_curves[i,]=mat[i,"beta0"]+gp_res$Z
  }
  save(xgrid,mat,Y_curves,file="data/2.10-plot.RData")
}
load("data/2.10-plot.RData")
library(svglite)
svglite("../markdown/figures/2.10.svg",width=4,height=4)
plot(NULL,xlim=c(0,1),ylim=c(10,32),xlab="x",ylab="y(x)")
for(i in 1:nrow(Y_curves)){
  lines(xgrid,Y_curves[i,])
  points(xgrid[seq(1,length(xgrid),length.out=20)],Y_curves[i,seq(1,length(xgrid),length.out=20)],pch=i)
}
dev.off()
```
